```python
restaurants = [
    {"name": "Karachi Biryani House", "location": "Karachi", "budget": "Low", "cuisine": "Pakistani"},
    {"name": "Al-Madina Grill", "location": "Karachi", "budget": "Medium", "cuisine": "BBQ"},
    {"name": "Ocean Breeze Cafe", "location": "Karachi", "budget": "High", "cuisine": "Continental"},
    {"name": "Lahori Tikka Corner", "location": "Lahore", "budget": "Low", "cuisine": "Pakistani"},
    {"name": "Spice Route", "location": "Lahore", "budget": "Medium", "cuisine": "Chinese"},
    {"name": "The Grand Diner", "location": "Lahore", "budget": "High", "cuisine": "Continental"},
    {"name": "Islamabad Fast Food", "location": "Islamabad", "budget": "Low", "cuisine": "Fast Food"},
    {"name": "Margalla Bistro", "location": "Islamabad", "budget": "Medium", "cuisine": "Italian"},
    {"name": "Skyline Restaurant", "location": "Islamabad", "budget": "High", "cuisine": "Continental"},
]


# ---- Agent Function ----
def food_recommendation_agent(location, budget, preference):
    """
    Percepts: location, budget, food preference (Sensors)
    Action: Recommended restaurant list (Actuators)
    """
    results = []

    for r in restaurants:
        if (r["location"].lower() == location.lower() and
            r["budget"].lower() == budget.lower() and
            r["cuisine"].lower() == preference.lower()):
            results.append(r)

    return results


# ---- Sensors: Getting input from the user ----
print("=== AI Food Recommendation Agent ===")
user_location = input("Enter your city (e.g., Karachi, Lahore, Islamabad): ")
user_budget = input("Enter your budget (Low / Medium / High): ")
user_preference = input("Enter food preference (e.g., Pakistani, BBQ, Chinese, Continental, Fast Food, Italian): ")

# ---- Actuators: Displaying the result ----
recommendations = food_recommendation_agent(user_location, user_budget, user_preference)

print("\n--- Recommendation Result ---")
if recommendations:
    print(f"We found {len(recommendations)} matching restaurant(s) for you:\n")
    for r in recommendations:
        print("Restaurant:", r["name"])
        print("Location:", r["location"])
        print("Budget:", r["budget"])
        print("Cuisine:", r["cuisine"])
        print("-" * 30)
else:
    print("Sorry, no restaurant found matching your preferences. Try different options.")
```
